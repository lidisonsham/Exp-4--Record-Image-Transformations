# **Image Scaling**

Write a Python program using OpenCV to perform image scaling on the **given image**.

The program should:

1. Scale the image to:
   - **0.5 times** its original size (downscaling)
   - **2 times** its original size (upscaling)
2. Use `cv2.resize` with appropriate scaling factors.
5. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Downscaled Image (0.5×)**  
   - **Upscaled Image (2×)**


```python
# Image Scaling
fx, fy = 0.5, 2.0  
scaled_image = cv2.resize(image, None, fx=fx, fy=fy, interpolation=cv2.INTER_LINEAR)
```

```python
# 2. Upscale to 2.0x its original size
upscaled = cv2.resize(img_rgb, None, fx=2.0, fy=2.0)
```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
h_matrix = np.float32([[1, 0.5, 0],
                       [0, 1, 0]])

# Vertical shearing matrix
v_matrix = np.float32([[1, 0, 0],
                       [0.5, 1, 0]])
```

```python
# Apply shearing
horizontal_shear = cv2.warpAffine(image, h_matrix, (600, 400))
vertical_shear = cv2.warpAffine(image, v_matrix, (600, 400))

```

```python

# Display images
plt.figure(figsize=(12,4))
```

```python

plt.subplot(1,3,1)
plt.imshow(image[:,:,::-1])
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(horizontal_shear[:,:,::-1])
plt.title("Horizontal Shear")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(vertical_shear[:,:,::-1])
plt.title("Vertical Shear")
plt.axis('off')

plt.show()
```

# **Image Reflection**

Write a Python program using OpenCV to perform image reflection on the **given image**.

The program should:

1. Generate the following reflected versions of the image:
   - **Horizontal Reflection** (flip left ↔ right)  
   - **Vertical Reflection** (flip top ↔ bottom)  
   - **Both-axis Reflection**  
2. Use `cv2.flip` with appropriate flip codes.
3. Display the following using a **2 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Reflected Image**  
   - **Vertically Reflected Image**  
   - **Both-axis Reflected Image**


```python
# Reflection using cv2.flip
horizontal_reflection = cv2.flip(image, 1)   # Left ↔ Right
vertical_reflection = cv2.flip(image, 0)     # Top ↔ Bottom
both_reflection = cv2.flip(image, -1)        # Both axes
```

```python
# Display images
plt.figure(figsize=(10, 10))
```

```python

# Original Image
plt.subplot(2, 2, 1)
plt.imshow(image[:, :, ::-1])
plt.title("Original Image")
plt.axis('off')

# Horizontal Reflection
plt.subplot(2, 2, 2)
plt.imshow(horizontal_reflection[:, :, ::-1])
plt.title("Horizontal Reflection")
plt.axis('off')

# Vertical Reflection
plt.subplot(2, 2, 3)
plt.imshow(vertical_reflection[:, :, ::-1])
plt.title("Vertical Reflection")
plt.axis('off')

# Both-axis Reflection
plt.subplot(2, 2, 4)
plt.imshow(both_reflection[:, :, ::-1])
plt.title("Both-axis Reflection")
plt.axis('off')

plt.tight_layout()
plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
# Get image size
(height, width) = image.shape[:2]
```

```python
# Find center of image
center = (width // 2, height // 2)
```

```python
# Rotation by 45 degrees
M1 = cv2.getRotationMatrix2D(center, 45, 1)
rotated_45 = cv2.warpAffine(image, M1, (width, height))
```

```python
# Rotation by 90 degrees
M2 = cv2.getRotationMatrix2D(center, 90, 1)
rotated_90 = cv2.warpAffine(image, M2, (width, height))
```

```python
# Display images
plt.figure(figsize=(12, 4))
```

```python
# Original Image
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')

# 45 Degree Rotation
plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(rotated_45, cv2.COLOR_BGR2RGB))
plt.title("45° Rotated Image")
plt.axis('off')

# 90 Degree Rotation
plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(rotated_90, cv2.COLOR_BGR2RGB))
plt.title("90° Rotated Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```
